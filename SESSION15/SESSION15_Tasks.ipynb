```python
import pandas as pd
from scipy import stats
import numpy as np
```

```python
#task1

data = {
    'customer_id': [f'CUST_{i:02d}' for i in range(1, 16)],
    'age_group': (
        ['18-25'] * 5 +
        ['26-35'] * 5 +
        ['36-50'] * 5
    ),
    'purchase_value': [
        450, 520, 480, 600, 550,   
        850, 920, 780, 890, 960,  
        1200, 1150, 1300, 1250, 1100  
    ]
}

df = pd.DataFrame(data)
df.to_csv('purchase_data.csv', index=False)

print("Saved 'purchase_data.csv' successfully.")
print(df)
```

```python
#task2

import pandas as pd
from scipy import stats

df = pd.read_csv('purchase_data.csv')

group_18_25 = df[df['age_group'] == '18-25']['purchase_value']
group_26_35 = df[df['age_group'] == '26-35']['purchase_value']
group_36_50 = df[df['age_group'] == '36-50']['purchase_value']


f_stat, p_val = stats.f_oneway(group_18_25, group_26_35, group_36_50)

print(f"Group Means:")
print(f"18-25: {group_18_25.mean():.2f}")
print(f"26-35: {group_26_35.mean():.2f}")
print(f"36-50: {group_36_50.mean():.2f}")
print("-" * 40)
print(f"F-statistic: {f_stat:.4f}")
print(f"p-value: {p_val:.8e}")
```

```python
#task3

interpretation = """
The one-way ANOVA yields a very large F-statistic and a p-value far below the standard significance threshold (p < 0.05).
Therefore, we reject the null hypothesis and conclude that there is a statistically significant difference in average purchase values across the three age groups.
Older customers (36-50) spend significantly more on average compared to the younger demographics (18-25 and 26-35).
"""

print("ANOVA Results Interpretation:")
print(interpretation.strip())
```

```python
#task4

df = pd.read_csv('purchase_data.csv')

grand_mean = df['purchase_value'].mean()
k = df['age_group'].nunique()  
N = len(df)                   

ssb = 0
for _, group in df.groupby('age_group'):
    n_i = len(group)
    mean_i = group['purchase_value'].mean()
    ssb += n_i * ((mean_i - grand_mean) ** 2)

df_between = k - 1
between_group_variance = ssb / df_between  

ssw = 0
for _, group in df.groupby('age_group'):
    mean_i = group['purchase_value'].mean()
    ssw += np.sum((group['purchase_value'] - mean_i) ** 2)

df_within = N - k
within_group_variance = ssw / df_within  

print(f"Between-Group Variance (MSB): {between_group_variance:,.2f}")
print(f"Within-Group Variance (MSW):  {within_group_variance:,.2f}")

print("\nExplanation:")
print("- Between-Group Variance measures how much the average spending differs from one age group to another.")
print("- Within-Group Variance measures the random individual spending fluctuations inside each single age group.")
```

```python
#task5

students = [1200, 1500, 1350, 1100, 1600, 1400, 1250]
professionals = [3800, 4200, 3950, 4500, 4100, 3700, 4300]
retirees = [2100, 2400, 2000, 2600, 2300, 2200, 2500]

f_stat, p_val = stats.f_oneway(students, professionals, retirees)

print("Zomato Customer Segment Analysis (One-Way ANOVA):")
print(f"Average Orders - Students:      ₹{np.mean(students):.2f}")
print(f"Average Orders - Professionals: ₹{np.mean(professionals):.2f}")
print(f"Average Orders - Retirees:      ₹{np.mean(retirees):.2f}")
print("-" * 50)
print(f"F-statistic: {f_stat:.4f}")
print(f"p-value:     {p_val:.8e}")

alpha = 0.05
if p_val < alpha:
    print("\nResult: Reject H0 — Monthly order spending differs significantly across the three customer segments.")
else:
    print("\nResult: Fail to reject H0 — No significant difference in monthly order spending across segments.")
```
