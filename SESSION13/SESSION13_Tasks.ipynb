```python
import numpy as np
from scipy import stats
```

```python
#task1

delivery_times = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]
target_mean = 30

t_stat, p_val = stats.ttest_1samp(delivery_times, target_mean)

sample_mean = np.mean(delivery_times)
print(f"Sample Mean: {sample_mean:.2f} mins")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.4f}")

alpha = 0.05
if p_val < alpha:
  print("Conclusion: Reject H0 — Delivery time is significantly different from 30 mins.")
else:
  print("Conclusion: Fail to reject H0 — No significant difference from 30 mins.")
```

```python
#task2

steps_before = [6200, 5800, 6500, 7100, 5900, 6300, 6800, 6000, 6400, 6700]
steps_after = [7500, 7200, 7800, 8100, 7000, 7600, 8300, 7400, 7900, 8000]

t_stat, p_val = stats.ttest_ind(steps_before, steps_after)

print(f"Mean Steps Before: {np.mean(steps_before):.1f}")
print(f"Mean Steps After: {np.mean(steps_after):.1f}")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.6f}")

alpha = 0.05
if p_val < alpha:
  print("Conclusion: Reject H0 — Daily steps significantly increased after using the app.")
else:
  print("Conclusion: Fail to reject H0 — No significant increase in steps.")
```

```python
#task3

screen_before = [6.5, 7.0, 5.8, 8.2, 6.0, 7.5, 6.8, 5.5, 7.2, 6.9]
screen_after = [4.8, 5.2, 4.5, 6.0, 4.9, 5.5, 5.0, 4.2, 5.3, 5.1]

t_stat, p_val = stats.ttest_rel(screen_before, screen_after)

print(f"Mean Screen Time Before: {np.mean(screen_before):.2f} hrs")
print(f"Mean Screen Time After: {np.mean(screen_after):.2f} hrs")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.8f}")

alpha = 0.05
if p_val < alpha:
  print("Conclusion: Reject H0 — Focus Mode significantly reduced screen time.")
else:
  print("Conclusion: Fail to reject H0 — No significant reduction in screen time.")
```

```python
#task4

zomato_ratings = [4.2, 3.8, 4.5, 4.0, 3.9, 4.1, 4.6, 3.7, 4.3, 4.2, 4.0, 4.4, 3.8, 4.1, 4.3]
swiggy_ratings = [4.0, 4.1, 3.9, 4.2, 3.8, 4.0, 4.3, 3.9, 4.1, 4.0, 4.2, 3.7, 4.0, 4.1, 3.9]

t_stat, p_val = stats.ttest_ind(zomato_ratings, swiggy_ratings)

print(f"Zomato Mean Rating: {np.mean(zomato_ratings):.2f}")
print(f"Swiggy Mean Rating: {np.mean(swiggy_ratings):.2f}")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.4f}")

print("\nWhy this test was chosen:")
print("An independent two-sample t-test (ttest_ind) was chosen because the ratings")
print("come from two separate, unrelated groups of users evaluating two different apps.")
```

```python
#task5


phonepe_txns = [4, 6, 5, 7, 3, 5, 6, 8, 4, 5, 6, 7]
claimed_benchmark = 5  

t_stat, p_val = stats.ttest_1samp(phonepe_txns, claimed_benchmark)

print("One-Sample t-Test on PhonePe Transactions:")
print(f"Sample Transactions: {phonepe_txns}")
print(f"Sample Mean: {sum(phonepe_txns) / len(phonepe_txns):.2f}")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.4f}")

alpha = 0.05
if p_val < alpha:
  print("Result: Significant difference from the benchmark.")
else:
  print("Result: No statistically significant difference from the benchmark of 5 transactions/day.")
```
