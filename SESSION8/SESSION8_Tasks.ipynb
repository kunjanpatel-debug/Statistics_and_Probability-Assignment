```python
import numpy as np
import scipy.stats as stats
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
```

```python
# Task 1: Bernoulli Trial Simulation

np.random.seed(42)
p_login = 0.85
n_simulations = 20

login_trials = np.random.binomial(n=1, p=p_login, size=n_simulations)

print("=" * 75)
print("TASK 1: Bernoulli Trial Simulation (Zomato Login)")
print("=" * 75)
print(f"Probability of Success (p): {p_login}")
print(f"Simulated Trials (1 = Success, 0 = Failure):\n{login_trials.tolist()}")
print(f"Total Successful Logins: {np.sum(login_trials)} / {n_simulations} ({np.mean(login_trials) * 100:.1f}%)\n")
```

```python
# Task 2: Binomial Distribution (Paytm UPI Transactions)

def binomial_upi_success(n, k, p):
  
    return stats.binom.pmf(k, n, p)

n_upi = 5
k_upi = 3
p_upi = 0.90

prob_exact_3 = binomial_upi_success(n_upi, k_upi, p_upi)

print("=" * 75)
print("TASK 2: Binomial Probability of Exactly 3 Successful UPI Payments")
print("=" * 75)
print(f"Total Attempts (n): {n_upi}")
print(f"Target Successes (k): {k_upi}")
print(f"Single Attempt Success Rate (p): {p_upi}")
print(f"Formula: P(X = 3) = 5C3 * (0.9)^3 * (0.1)^2 = 10 * 0.729 * 0.01 = 0.0729")
print(f"Calculated Probability: {prob_exact_3:.4f} ({prob_exact_3 * 100:.2f}%)\n")
```

```python
# Task 3: Poisson Distribution (Spotify Playlist Creation Rate)

lam_spotify = 4  
k_spotify = 6    

prob_poisson_6 = stats.poisson.pmf(k_spotify, lam_spotify)

print("=" * 75)
print("TASK 3: Poisson Probability of Exactly 6 Playlists Created in a Minute")
print("=" * 75)
print(f"Average Creation Rate (λ): {lam_spotify} playlists/min")
print(f"Target Count (k): {k_spotify} playlists")
print(f"Formula: P(X = 6) = (e^(-4) * 4^6) / 6! = (0.0183156 * 4096) / 720 ≈ 0.1042")
print(f"Calculated Probability: {prob_poisson_6:.4f} ({prob_poisson_6 * 100:.2f}%)\n")
```

```python

# Task 4: Simulate & Plot 12 Hours of Reviews via Poisson (Flipkart)

np.random.seed(42)
lam_reviews = 10
hours_count = 12

simulated_reviews = np.random.poisson(lam_reviews, size=hours_count)

plt.figure(figsize=(9, 4.8))
hour_labels = [f"Hr {i+1}" for i in range(hours_count)]
bars = plt.bar(hour_labels, simulated_reviews, color='#2874F0', edgecolor='black', width=0.55, alpha=0.88)
plt.axhline(lam_reviews, color='#E23744', linestyle='--', linewidth=2, label=f'Expected Mean Rate (λ = {lam_reviews})')

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, yval + 0.3, str(yval), ha='center', va='bottom', fontsize=9.5, fontweight='bold')

plt.title('Flipkart Product Reviews Received per Hour (12-Hour Poisson Simulation)', fontsize=12, fontweight='bold', pad=12)
plt.xlabel('Observation Hour', fontsize=10.5)
plt.ylabel('Number of Reviews Received', fontsize=10.5)
plt.ylim(0, max(simulated_reviews) + 4)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(frameon=True, facecolor='#FAFAFA', edgecolor='#CCCCCC')

plt.tight_layout()
plt.savefig('session8_task4_poisson_simulation.png', dpi=150)
plt.close()

print("=" * 75)
print("TASK 4: Simulated 12-Hour Poisson Reviews (Flipkart)")
print("=" * 75)
print(f"Simulated Review Counts: {simulated_reviews.tolist()}")
print("Chart generated and saved as session8_task4_poisson_simulation.png\n")
```
