```python
import pandas as pd
from scipy import stats
```

```python
#task1

data = {
    'Paytm': [6, 4, 2],
    'PhonePe': [5, 6, 1],
    'Credit Card': [3, 2, 1]
}
apps = ['Zomato', 'Swiggy', 'Uber Eats']

contingency_table = pd.DataFrame(data, index=apps)
contingency_table['Total'] = contingency_table.sum(axis=1)

# Add column totals row
totals_row = contingency_table.sum(axis=0)
totals_row.name = 'Total'
contingency_table = pd.concat([contingency_table, pd.DataFrame([totals_row])])

print("Contingency Table (App vs Payment Method):")
print(contingency_table)
```

```python
#task2

data = {
    'Males': [12, 8, 10],
    'Females': [18, 7, 15]
}
platforms = ['Spotify', 'Gaana', 'YouTube Music']
observed_df = pd.DataFrame(data, index=platforms)

row_totals = observed_df.sum(axis=1)
col_totals = observed_df.sum(axis=0)
grand_total = observed_df.values.sum()

expected_df = pd.DataFrame(index=platforms, columns=['Males', 'Females'])
for app in platforms:
    for gender in ['Males', 'Females']:
        expected_val = (row_totals[app] * col_totals[gender]) / grand_total
        expected_df.loc[app, gender] = round(expected_val, 2)

print("Observed Counts:")
print(observed_df)
print("\nExpected Counts (Assuming Independence):")
print(expected_df)
```

```python
#task3


observed = [
    [6, 5, 3],  # Zomato
    [4, 6, 2],  # Swiggy
    [2, 1, 1]   # Uber Eats
]

# Chi-Square Test of Independence
chi2_stat, p_val, dof, expected = stats.chi2_contingency(observed)

print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"Degrees of Freedom: {dof}")
print(f"p-value: {p_val:.4f}")
print("\nExpected Frequencies:")
print(pd.DataFrame(expected, index=['Zomato', 'Swiggy', 'Uber Eats'], columns=['Paytm', 'PhonePe', 'Credit Card']).round(2))

alpha = 0.05
if p_val < alpha:
    print("\nConclusion: Reject H0 — Significant association between food delivery app choice and payment method.")
else:
    print("\nConclusion: Fail to reject H0 — No significant association; app choice and payment method are independent.")
```

```python
#task4

observed_counts = [18, 12]                      
expected_counts = [30 * 0.70, 30 * 0.30]        

chi2_stat, p_val = stats.chisquare(f_obs=observed_counts, f_exp=expected_counts)

print(f"Observed Counts: {observed_counts}")
print(f"Expected Counts: {expected_counts}")
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"p-value: {p_val:.4f}")

alpha = 0.05
if p_val < alpha:
    print("Conclusion: Reject H0 — The observed data significantly differs from the company's 70% claim.")
else:
    print("Conclusion: Fail to reject H0 — The survey results are consistent with the company's 70% claim.")
```
