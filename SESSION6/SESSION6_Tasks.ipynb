```python
die_faces = [1, 2, 3, 4, 5, 6]
coin_sides = ["H", "T"]

sample_space = [(d, c) for d in die_faces for c in coin_sides]
print("Sample Space:", sample_space)
print("Total Outcomes:", len(sample_space))  # 12
```

```python
deck = list(range(1, 11))
even_cards = [c for c in deck if c % 2 == 0]
p_even = len(even_cards) / len(deck)

print(f"P(Even) = {len(even_cards)} / {len(deck)} = {p_even}")  # 0.5
```

```python
total_songs = 8
arijit_songs = 3

p_arijit = arijit_songs / total_songs
p_not_arijit = 1 - p_arijit

print(
    f"P(NOT Arijit Singh) = 1 - {p_arijit} = {p_not_arijit}"
)  # 5/8 = 0.625 (62.5%)
```

```python
p_pizza = 0.3
p_burger = 0.4
p_both = 0.1

p_pizza_or_burger = p_pizza + p_burger - p_both
print(f"P(Pizza or Burger) = {p_pizza_or_burger:.2f}")  # 0.60
```

```python
p_cart = 0.2
p_cart_given_interest = 0.5

print("=" * 75)
print("TASK 5: Conditional Probability Interpretation")
print("=" * 75)
print(f"Prior Probability P(Cart) = {p_cart}")
print(f"Conditional / Posterior Probability P(Cart | Interested in Mobiles) = {p_cart_given_interest}")
explanation = (
    "Interpretation:\n"
    "Unconditionally, any random visitor browsing electronics has only a 20% baseline chance of adding a phone to cart.\n"
    "However, when new contextual evidence is observed—specifically that the user explicitly searches for or explores mobile phones—"
    "the probability updates from 0.2 to 0.5 (a 2.5x increase).\n"
    "Conditional probability P(A|B) quantifies how fresh evidence (B) narrows down the sample space and refines our belief about event (A)."
)
print(explanation)
```
