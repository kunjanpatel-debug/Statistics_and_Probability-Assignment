```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#task1

np.random.seed(42)
population = np.random.exponential(scale=50, size=50000)

sample_size = 30
num_samples = 1000
sample_means = [
    np.mean(np.random.choice(population, size=sample_size))
    for _ in range(num_samples)
]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(population, bins=40, color='coral', edgecolor='black', alpha=0.7)
plt.title('Original Population (Highly Skewed Likes)')
plt.xlabel('Likes per Post')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(
    sample_means, bins=30, color='dodgerblue', edgecolor='black', alpha=0.7
)
plt.title('Distribution of Sample Means (n = 30)')
plt.xlabel('Sample Mean Likes')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

print(f'Population Mean: {np.mean(population):.2f}')
print(f'Mean of 1,000 Sample Means: {np.mean(sample_means):.2f}')
```

```python
#task2

explanation = """
Why the sampling distribution becomes normal (Zomato Example):
Individual Zomato order values are heavily right-skewed—most orders are modest everyday meals (around ₹150–₹350), with a long tail of rare, expensive party orders (₹3,000+).

When you take a random sample of 30 orders and calculate their average, extreme low values and extreme high values pull in opposite directions and neutralize one another. The sample mean will naturally fall near the central average most of the time, while an entire sample consisting purely of ₹3,000+ orders is statistically almost impossible. Because balance is the most probable outcome, repeating this 1,000 times creates a symmetrical, bell-shaped distribution centered at the true population mean.
"""

print(explanation.strip())
```

```python
#task4

np.random.seed(42)
population = np.random.exponential(scale=50, size=50000)

num_samples = 1000

means_n5 = [
    np.mean(np.random.choice(population, size=5)) for _ in range(num_samples)
]
means_n30 = [
    np.mean(np.random.choice(population, size=30)) for _ in range(num_samples)
]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(means_n5, bins=30, color='tomato', edgecolor='black', alpha=0.7)
plt.title('Sample Means with n = 5 (Small Sample)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(means_n30, bins=30, color='seagreen', edgecolor='black', alpha=0.7)
plt.title('Sample Means with n = 30 (Standard Sample)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

print('Observations when reducing sample size from n=30 to n=5:')
print(
    '1. Normality: n=5 still retains visible right-skewness and is not yet a'
    ' smooth bell curve, whereas n=30 is symmetrical.'
)
print(
    f'2. Spread/Variance: n=5 has a much wider spread (Std ='
    f' {np.std(means_n5):.2f}) than n=30 (Std = {np.std(means_n30):.2f})'
    ' because Standard Error decreases by 1/sqrt(n).'
)
```

```python
#task5

np.random.seed(101)
exp_population = np.random.exponential(scale=10, size=20000)

sample_sizes = [2, 10, 30, 100]
plt.figure(figsize=(12, 8))

for idx, n in enumerate(sample_sizes, 1):
  sample_means = [
      np.mean(np.random.choice(exp_population, size=n)) for _ in range(1000)
  ]
  plt.subplot(2, 2, idx)
  plt.hist(
      sample_means,
      bins=30,
      density=True,
      color='#6A0DAD',
      edgecolor='black',
      alpha=0.65,
  )
  plt.title(f'Sample Size n = {n}')
  plt.xlabel('Sample Mean')
  plt.ylabel('Density')

plt.tight_layout()
plt.show()

print('Observation:')
print(
    'At n=2, the distribution remains heavily skewed like the raw exponential'
    ' data.'
)
print(
    'As n increases to 10, 30, and 100, the shape transforms into a clear,'
    ' tight bell curve with rapidly narrowing variance, visually confirming'
    ' the Central Limit Theorem.'
)
```
