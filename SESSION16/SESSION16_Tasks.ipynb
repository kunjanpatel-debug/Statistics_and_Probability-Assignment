```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#task1

songs_data = {
    'song_name': [f'Track_{i:02d}' for i in range(1, 13)],
    'danceability': [0.82, 0.75, 0.65, 0.90, 0.55, 0.78, 0.60, 0.85, 0.70, 0.92, 0.50, 0.88],
    'energy':       [0.70, 0.68, 0.58, 0.85, 0.45, 0.72, 0.50, 0.80, 0.65, 0.88, 0.40, 0.82],
    'popularity':   [85, 78, 62, 92, 50, 80, 58, 88, 72, 95, 45, 90]
}

df_songs = pd.DataFrame(songs_data)

pearson_corr = df_songs['danceability'].corr(df_songs['popularity'], method='pearson')

print("Songs Dataset (First 5 Rows):")
print(df_songs.head())
print("-" * 50)
print(f"Pearson Correlation (Danceability vs. Popularity): {pearson_corr:.4f}")
```

```python
#task2

ipl_data = {
    'player': ['Virat Kohli', 'Rohit Sharma', 'KL Rahul', 'Shubman Gill', 
               'Rishabh Pant', 'Hardik Pandya', 'Suryakumar Yadav', 'Rinku Singh'],
    'total_runs': [7500, 6600, 4600, 3200, 3400, 2500, 3600, 950],
    'followers_millions': [270, 42, 17, 15, 12, 35, 16, 5]
}

df_ipl = pd.DataFrame(ipl_data)

corr_val, p_val = stats.spearmanr(df_ipl['total_runs'], df_ipl['followers_millions'])

print("IPL Players Data:")
print(df_ipl[['player', 'total_runs', 'followers_millions']])
print("-" * 50)
print(f"Spearman Correlation: {corr_val:.4f}")
print(f"p-value: {p_val:.4f}")

print("\nInterpretation:")
if corr_val > 0.5:
    print("Yes, there is a strong positive monotonic relationship (rho = 0.74).")
    print("Higher total runs generally correspond to higher follower counts, though brand popularity and seniority also influence social reach.")
```

```python
#task3

flipkart_data = {
    'price': [12000, 25000, 8000, 45000, 15000, 30000, 5000, 60000, 18000, 22000],
    'user_rating': [4.2, 4.5, 3.8, 4.6, 4.1, 4.4, 3.7, 4.7, 4.0, 4.3],
    'num_reviews': [1500, 3200, 800, 5000, 2100, 3800, 450, 6200, 2400, 2900]
}

df_products = pd.DataFrame(flipkart_data)

correlation_matrix = df_products.corr()

print("Flipkart Products Correlation Matrix:")
print(correlation_matrix.round(4))
```

```python
#task4

import matplotlib.pyplot as plt


plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix, 
    annot=True, 
    cmap='coolwarm', 
    fmt='.2f', 
    linewidths=1, 
    cbar=True
)

plt.title('Flipkart Products Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```
