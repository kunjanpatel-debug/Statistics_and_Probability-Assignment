```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python


np.random.seed(42)
mu = 500
sigma = 100
samples = np.random.normal(loc=mu, scale=sigma, size=1000)

plt.figure(figsize=(9, 5))
count, bins, _ = plt.hist(samples, bins=30, density=True, alpha=0.65, color='#007ACC', edgecolor='black')
pdf_curve = stats.norm.pdf(bins, loc=mu, scale=sigma)
plt.plot(bins, pdf_curve, color='#DC3545', linewidth=2.5, label='Theoretical PDF')
plt.axvline(mu, color='black', linestyle='--', linewidth=1.5, label=f'Mean (μ = {mu})')

plt.title('Normal Distribution Simulation (1,000 Samples, μ = 500, σ = 100)')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

print(f"Sample Mean: {np.mean(samples):.2f}, Sample Std: {np.std(samples):.2f}")
```

```python
#task3

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

z_values = np.linspace(-4, 4, 1000)
pdf_standard = stats.norm.pdf(z_values, loc=0, scale=1)

plt.figure(figsize=(9, 5))
plt.plot(z_values, pdf_standard, color='#1F4E79', linewidth=2.5, label='Standard Normal N(0, 1)')

z_shade = np.linspace(-1, 1, 500)
pdf_shade = stats.norm.pdf(z_shade, loc=0, scale=1)
plt.fill_between(z_shade, pdf_shade, color='#4CAF50', alpha=0.45, label='Area: -1 ≤ Z ≤ 1')

plt.axvline(0, color='black', linestyle='--', linewidth=1.2, label='Mean (Z = 0)')
plt.axvline(-1, color='#DC3545', linestyle=':', linewidth=1.5)
plt.axvline(1, color='#DC3545', linestyle=':', linewidth=1.5)

plt.title('Standard Normal Curve: Area (-1 ≤ Z ≤ 1)')
plt.xlabel('Z-Score')
plt.ylabel('Density')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

print(f"Shaded Area: {(stats.norm.cdf(1) - stats.norm.cdf(-1)) * 100:.2f}%")
```

```python
#task4

from scipy.stats import norm

mu_delivery = 30
sigma_delivery = 5
target_time = 25

prob_less_25 = norm.cdf(target_time, loc=mu_delivery, scale=sigma_delivery)
z_val = (target_time - mu_delivery) / sigma_delivery

print(f"Z-score: {z_val:.2f}")
print(f"P(Delivery < 25 mins): {prob_less_25:.4f}")
```
