```python
p_headphones = 0.30
p_smartphone_given_headphones = 0.40

p_both_t2 = p_headphones * p_smartphone_given_headphones

p_calc = p_both_t2 / p_headphones

print("=" * 75)
print("TASK 2: Flipkart User Purchase Probability")
print("=" * 75)
print(f"P(Headphones) = {p_headphones:.2f}")
print(f"P(Headphones ∩ Smartphone) = {p_both_t2:.2f}")
print(f"Formula: P(Smartphone | Headphones) = P(Smartphone ∩ Headphones) / P(Headphones)")
print(f"P(Smartphone | Headphones) = {p_both_t2:.2f} / {p_headphones:.2f} = {p_calc:.2f} ({p_calc * 100:.1f}%)\n")
```

```python
def bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email):

    return (prob_email_given_churn * prior_churn) / prob_email

prior_c = 0.15
prob_e_given_c = 0.60
prob_e = 0.25

result_churn = bayes_churn_probability(prior_c, prob_e_given_c, prob_e)

print("=" * 75)
print("TASK 3: Bayes' Theorem Churn Prediction Function")
print("=" * 75)
print(f"Prior Probability of Churn P(Churn): {prior_c}")
print(f"Likelihood P(Email | Churn): {prob_e_given_c}")
print(f"Marginal Probability P(Email): {prob_e}")
print(f"Calculated P(Churn | Email) = ({prob_e_given_c} * {prior_c}) / {prob_e} = {result_churn:.4f} ({result_churn * 100:.2f}%)\n")
```

```python
p_f = 0.02          
p_not_f = 1 - p_f   

p_a_given_f = 0.90      
p_a_given_not_f = 0.05  

p_a = (p_a_given_f * p_f) + (p_a_given_not_f * p_not_f)

p_f_given_a = (p_a_given_f * p_f) / p_a

print("=" * 75)
print("TASK 4: Wallet Fraud Detection (Bayes' Theorem)")
print("=" * 75)
print(f"Base Rate of Fraud P(F): {p_f}")
print(f"True Positive Rate P(A | F): {p_a_given_f}")
print(f"False Positive Rate P(A | F'): {p_a_given_not_f}")
print(f"Total Alert Probability P(A) = ({p_a_given_f} * {p_f}) + ({p_a_given_not_f} * {p_not_f}) = {p_a:.4f}")
print(f"Posterior Probability P(F | A) = ({p_a_given_f} * {p_f}) / {p_a:.4f} = {p_f_given_a:.4f} ({p_f_given_a * 100:.2f}%)\n")
```
