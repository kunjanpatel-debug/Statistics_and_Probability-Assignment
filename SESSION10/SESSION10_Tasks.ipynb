```python
import numpy as np
import pandas as pd
```

```python
import random

population_users = [f"SPOT_USER_{i:03d}" for i in range(1, 201)]

random.seed(42)
sample_users = random.sample(population_users, 20)

print(f"Total Users in Population: {len(population_users)}")
print(f"Sample Size Selected: {len(sample_users)}")
print("\nSelected User IDs for Feedback Survey:")
print(sample_users)
```

```python
#task2

np.random.seed(42)

cities = ['Ahmedabad', 'Surat', 'Vadodara']
city_probabilities = [0.50, 0.30, 0.20]

orders_df = pd.DataFrame({
    'Order_ID': [f"OD_{i:04d}" for i in range(1, 1001)],
    'City': np.random.choice(cities, size=1000, p=city_probabilities),
})

target_sample_size = 60

samples_list = []
for city, group in orders_df.groupby('City'):
  quota = int(round((len(group) / len(orders_df)) * target_sample_size))
  samples_list.append(group.sample(n=quota, random_state=42))

stratified_sample = pd.concat(samples_list).reset_index(drop=True)

print("Population Distribution by City:")
print(orders_df['City'].value_counts())
print("\nStratified Sample Distribution by City (Total = 60):")
print(stratified_sample['City'].value_counts())
print("\nFirst 5 Stratified Sample Records:")
print(stratified_sample.head())
```

```python
#task3

reviews_data = [
    {"Review_ID": f"REV_{i:03d}", "Rating": (i % 5) + 1, "Text": f"Food quality review number {i}"}
    for i in range(1, 501)
]

def systematic_sampling(data_list, k=10, start_index=0):
    """
    Selects every k-th item starting from start_index.
    """
    return [data_list[i] for i in range(start_index, len(data_list), k)]

sample_reviews = systematic_sampling(reviews_data, k=10, start_index=0)
sample_df = pd.DataFrame(sample_reviews)

print(f"Total Reviews in Population: {len(reviews_data)}")
print(f"Sampling Interval (k): Every 10th review")
print(f"Total Sampled Reviews: {len(sample_reviews)}")
print("\nSampled Review IDs (First 10 shown):")
print(sample_df['Review_ID'].head(10).tolist())
```

```python
#task4

population_size = 10000
assumed_sigma = 8.0  
confidence_level = 1.96 
sample_sizes = [50, 200, 500]

results = []
for n in sample_sizes:
    se = (assumed_sigma / np.sqrt(n)) * np.sqrt((population_size - n) / (population_size - 1))
    margin_of_error = confidence_level * se
    results.append({
        "Sample Size (n)": n,
        "Sampling Ratio (%)": f"{(n / population_size) * 100:.1f}%",
        "Standard Error (mins)": round(se, 4),
        "95% Margin of Error (± mins)": round(margin_of_error, 2)
    })

comparison_df = pd.DataFrame(results)
print("Impact of Sample Size on Estimate Precision:")
print(comparison_df.to_string(index=False))

print("\nKey Takeaway:")
print("- Increasing sample size from 50 to 200 cuts the margin of error in half (from ±2.21 to ±1.10 mins).")
print("- At n = 500, the estimate stays within ±0.68 minutes of the true population mean at 95% confidence.")
```

```python
#task5


np.random.seed(101)

teams = ['SRH', 'MI', 'RCB', 'KKR', 'GT']
team_weights = [0.35, 0.25, 0.20, 0.12, 0.08]

fans_df = pd.DataFrame({
    'Fan_ID': [f"FAN_{i:04d}" for i in range(1, 801)],
    'Favorite_Team': np.random.choice(teams, size=800, p=team_weights),
    'Age': np.random.randint(18, 50, size=800)
})

target_sample = 40
team_counts = fans_df['Favorite_Team'].value_counts()
team_sample_quotas = (team_counts / len(fans_df) * target_sample).round().astype(int)

while team_sample_quotas.sum() != target_sample:
    diff = target_sample - team_sample_quotas.sum()
    team_sample_quotas[team_sample_quotas.index[0]] += diff

sampled_fans_list = []
for team, quota in team_sample_quotas.items():
    team_group = fans_df[fans_df['Favorite_Team'] == team]
    sampled_fans_list.append(team_group.sample(n=quota, random_state=101))

stratified_fans_sample = pd.concat(sampled_fans_list).reset_index(drop=True)

print("Population Fan Count by Team:")
print(team_counts)
print("\nStratified Sample Quotas (Total = 40):")
print(team_sample_quotas)
print("\nSampled Fans Breakdown:")
print(stratified_fans_sample['Favorite_Team'].value_counts())
print("\nFirst 5 Sampled Fans:")
print(stratified_fans_sample.head())
```
